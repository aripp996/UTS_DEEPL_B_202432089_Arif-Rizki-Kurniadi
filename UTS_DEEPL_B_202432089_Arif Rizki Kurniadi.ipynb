Meskipun memiliki enam kolom atribut metadata, pemodelan *Deep Learning* seperti LSTM umumnya hanya memanipulasi kolom 'text' sebagai fitur masukan (X) dan 'target' sebagai label keluaran (Y) guna melatih algoritma dalam memecahkan pola semantik dari bahasa media sosial yang sangat bising dan tidak baku.

  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>
</div>

```python
#202432089_ArifRizkiKurniadi
import re
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from torch.nn.utils.rnn import pack_padded_sequence

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Reproducibility & device
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device yang digunakan:', device)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432089_ArifRizkiKurniadi
# 1. Muat dataset Sentiment140 dari Kaggle (kagglehub; dataset publik)
import kagglehub
path = kagglehub.dataset_download('kazanova/sentiment140')
csv_path = os.path.join(path, 'training.1600000.processed.noemoticon.csv')

df = pd.read_csv(
    csv_path, encoding='latin-1', header=None,
    names=['target', 'ids', 'date', 'flag', 'user', 'text']
)[['target', 'text']]
print('Ukuran dataset asli:', df.shape)

# Ambil sampel seimbang agar training di Colab tidak terlalu lama
# (naikkan SAMPLE_PER_CLASS, maksimal 800000, untuk hasil lebih akurat)
SAMPLE_PER_CLASS = 100_000
df = df.groupby('target').sample(n=SAMPLE_PER_CLASS, random_state=SEED).reset_index(drop=True)

# Label biner: 0 = negatif, 4 -> 1 = positif
df['label'] = (df['target'] == 4).astype(int)
print(df['label'].value_counts())

# 2. Pembersihan data & pra-pemrosesan
def clean_text(t):
    t = t.lower()                                # case folding
    t = re.sub(r'&\w+;', ' ', t)                 # HTML entity (&amp;, &quot;)
    t = re.sub(r'http\S+|www\.\S+', ' ', t)      # URL
    t = re.sub(r'@\w+', ' ', t)                  # mention
    t = t.replace('#', '')                       # hashtag -> kata biasa
    t = re.sub(r"[^a-z\s']", ' ', t)             # sisakan huruf & apostrof (don't, can't)
    t = re.sub(r'\s+', ' ', t).strip()
    return t

df['clean'] = df['text'].apply(clean_text)
df['tokens'] = df['clean'].str.split()           # tokenisasi
df = df[df['tokens'].str.len() > 0].reset_index(drop=True)
print('\nContoh hasil pembersihan:')
print(df[['text', 'clean']].head(3).to_string())

# Split data latih & uji (80:20, stratified)
train_df, test_df = train_test_split(
    df, test_size=0.2, random_state=SEED, stratify=df['label']
)

# Vocabulary mapping (hanya dari data latih agar tidak terjadi data leakage)
MAX_VOCAB = 20000
PAD_IDX, UNK_IDX = 0, 1
counter = Counter(tok for toks in train_df['tokens'] for tok in toks)
vocab = {'<pad>': PAD_IDX, '<unk>': UNK_IDX}
for word, _ in counter.most_common(MAX_VOCAB - 2):
    vocab[word] = len(vocab)
VOCAB_SIZE = len(vocab)
print('\nUkuran vocabulary:', VOCAB_SIZE)

# Encoding + padding (post-padding) ke panjang tetap
MAX_LEN = 40
def encode(tokens):
    ids = [vocab.get(tok, UNK_IDX) for tok in tokens][:MAX_LEN]
    return ids + [PAD_IDX] * (MAX_LEN - len(ids))

X_train_np = np.array([encode(t) for t in train_df['tokens']])
X_test_np = np.array([encode(t) for t in test_df['tokens']])

# 3. Konversi menjadi PyTorch Tensor
X_train = torch.tensor(X_train_np, dtype=torch.long)
y_train = torch.tensor(train_df['label'].values, dtype=torch.float32)
X_test = torch.tensor(X_test_np, dtype=torch.long)
y_test = torch.tensor(test_df['label'].values, dtype=torch.float32)
print('X_train:', X_train.shape, '| X_test:', X_test.shape)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432089_ArifRizkiKurniadi
class SequenceModel(nn.Module):
    def __init__(self, vocab_size, embed_dim=128, hidden_size=128,
                 num_layers=1, dropout=0.4, pad_idx=0):
        super(SequenceModel, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=pad_idx)
        self.lstm = nn.LSTM(
            embed_dim, hidden_size, num_layers=num_layers, batch_first=True,
            bidirectional=True, dropout=dropout if num_layers > 1 else 0.0
        )
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Linear(hidden_size * 2, 1)   # 1 logit -> sentimen positif/negatif

    def forward(self, x):
        # panjang asli tiap kalimat, agar token <pad> tidak ikut diproses LSTM
        lengths = (x != 0).sum(dim=1).clamp(min=1).cpu()
        emb = self.dropout(self.embedding(x))
        packed = pack_padded_sequence(emb, lengths, batch_first=True, enforce_sorted=False)
        _, (h_n, _) = self.lstm(packed)
        # gabungkan hidden state terakhir arah maju & mundur
        h = torch.cat((h_n[-2], h_n[-1]), dim=1)
        return self.fc(self.dropout(h)).squeeze(1)  # logit
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432089_ArifRizkiKurniadi
# 1. Inisiasi hyperparameter
epochs = 5            # 50 epoch tidak diperlukan untuk data teks sebesar ini (rawan overfitting)
learning_rate = 0.001
batch_size = 256
embed_dim = 128
hidden_size = 128

# 2. Inisiasi model, loss function, dan optimizer
model = SequenceModel(VOCAB_SIZE, embed_dim=embed_dim, hidden_size=hidden_size).to(device)
criterion = nn.BCEWithLogitsLoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=batch_size, shuffle=True)
test_loader = DataLoader(TensorDataset(X_test, y_test), batch_size=batch_size)

def evaluate(loader):
    model.eval()
    total_loss, preds, actuals = 0.0, [], []
    with torch.no_grad():
        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)
            logits = model(xb)
            total_loss += criterion(logits, yb).item() * xb.size(0)
            preds.extend((torch.sigmoid(logits) >= 0.5).int().cpu().tolist())
            actuals.extend(yb.int().cpu().tolist())
    return total_loss / len(loader.dataset), np.array(preds), np.array(actuals)

# 3. Riwayat loss untuk visualisasi
train_losses, test_losses, test_accs = [], [], []

# 4. Training loop
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        loss = criterion(model(xb), yb)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)   # cegah exploding gradient
        optimizer.step()
        running_loss += loss.item() * xb.size(0)

    train_loss = running_loss / len(train_loader.dataset)
    test_loss, p, a = evaluate(test_loader)
    train_losses.append(train_loss)
    test_losses.append(test_loss)
    test_accs.append(accuracy_score(a, p))
    print(f'Epoch {epoch+1:02d}/{epochs} | Train Loss: {train_loss:.4f} | '
          f'Test Loss: {test_loss:.4f} | Test Acc: {test_accs[-1]:.4f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432089_ArifRizkiKurniadi
# 1. Plot Grafik Penurunan Loss Training terhadap Epoch
plt.figure(figsize=(10, 4))
plt.plot(range(1, epochs + 1), train_losses, color='#114F6A', linewidth=2, marker='o', label='Training Loss')
plt.plot(range(1, epochs + 1), test_losses, color='#1FA4B1', linewidth=2, marker='s', linestyle='--', label='Test Loss')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.xticks(range(1, epochs + 1))
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# 2. Perbandingan Data Aktual vs Hasil Prediksi Model
_, y_pred, y_true = evaluate(test_loader)
print(f'Akurasi Data Uji: {accuracy_score(y_true, y_pred):.4f}\n')
print(classification_report(y_true, y_pred, target_names=['Negatif', 'Positif'], digits=4))

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# (a) Distribusi kelas: aktual vs prediksi
labels = ['Negatif', 'Positif']
actual_counts = [np.sum(y_true == 0), np.sum(y_true == 1)]
pred_counts = [np.sum(y_pred == 0), np.sum(y_pred == 1)]
x = np.arange(2)
axes[0].bar(x - 0.2, actual_counts, 0.4, color='#114F6A', label='Aktual')
axes[0].bar(x + 0.2, pred_counts, 0.4, color='#1FA4B1', label='Prediksi')
axes[0].set_xticks(x)
axes[0].set_xticklabels(labels)
axes[0].set_title('Distribusi Kelas: Aktual vs Prediksi')
axes[0].set_ylabel('Jumlah Tweet')
axes[0].legend()
axes[0].grid(True, axis='y', linestyle='--', alpha=0.5)

# (b) Confusion matrix
cm = confusion_matrix(y_true, y_pred)
im = axes[1].imshow(cm, cmap='GnBu')
axes[1].set_xticks([0, 1]); axes[1].set_xticklabels(labels)
axes[1].set_yticks([0, 1]); axes[1].set_yticklabels(labels)
axes[1].set_xlabel('Prediksi'); axes[1].set_ylabel('Aktual')
axes[1].set_title('Confusion Matrix')
for i in range(2):
    for j in range(2):
        axes[1].text(j, i, f'{cm[i, j]:,}', ha='center', va='center',
                     color='white' if cm[i, j] > cm.max() / 2 else '#114F6A', fontweight='bold')
plt.colorbar(im, ax=axes[1], fraction=0.046)
plt.tight_layout()
plt.show()

# 3. Uji coba prediksi pada kalimat baru
def predict_sentiment(text):
    model.eval()
    ids = torch.tensor([encode(clean_text(text).split())], dtype=torch.long).to(device)
    with torch.no_grad():
        prob = torch.sigmoid(model(ids)).item()
    return ('Positif' if prob >= 0.5 else 'Negatif'), prob

for kalimat in ["I love this so much, best day ever!",
                "this is the worst service, i hate it",
                "I am not happy with this at all"]:
    label, prob = predict_sentiment(kalimat)
    print(f'{label:8s} (p_positif={prob:.3f}) | {kalimat}')
```
